##Modalities Combination

```python
import plotly.graph_objects as go
import pandas as pd

# === Output file name ===
file_name = "fig_distribution_of_included_studies_by_modality_combination.pdf"

# === Data ===
modality = [
    "Two", "Two", "Two", "Two", "Two", "Two", "Two",
    "Three", "Three", "Three", "Three", "Three", "Three", "Three", "Three",
    "Four", "Four", "Four"
]

modality_combination = [
    "Image + Text",
    "Text + Video",
    "Time series + Text",
    "Images + Genomic data",
    "Audio + Text",
    "Structured measurements + Text",
    "Binary representations + Image",
    "Audio + Text + Video",
    "Audio + Image + Text",
    "Image + Text + Video",
    "Image + Text + Social context",
    "3D shapes + Image + Text",
    "Audio + MIDI + Text",
    "Text + Image + Category data",
    "Genomic data + Images + Wearable sensor time series",
    "Audio + Image + Text + Video",
    "Image + Text + Video + Robotic actions",
    "Audio + Image + Laser position + Laser power measurements"
]

studies = [29, 4, 1, 1, 1, 1, 1, 8, 1, 1, 1, 1, 1, 1, 1, 3, 1, 1]

# === Create DataFrame ===
df = pd.DataFrame({
    "Modality": modality,
    "Modality Combination": modality_combination,
    "Studies": studies
})

# === Build hierarchy ===
root_label = "<b>Modality Combination</b>"
total_studies = df["Studies"].sum()

labels = [root_label]
parents = [""]
values = [total_studies]
text_template = [f"<b>{root_label}</b><br>n={total_studies}, 100%"]

# Group by modality
for mod in ["Two", "Three", "Four"]:
    group_df = df[df["Modality"] == mod].sort_values("Studies", ascending=False)
    parent_label = f"{mod} Modalities"
    parent_total = group_df["Studies"].sum()

    # Parent node
    labels.append(parent_label)
    parents.append(root_label)
    values.append(parent_total)
    text_template.append(
        f"<b>{parent_label}</b><br>n={parent_total}, {parent_total/total_studies:.1%}"
    )

    # Child nodes
    for _, row in group_df.iterrows():
        labels.append(row["Modality Combination"])
        parents.append(parent_label)
        values.append(row["Studies"])
        text_template.append(
            f"<b>{row['Modality Combination']}</b><br>n={row['Studies']}, {row['Studies']/total_studies:.2%}"
        )

# === Assign colours by modality group ===
group_colors = {
    "Two Modalities": "royalblue",
    "Three Modalities": "seagreen",
    "Four Modalities": "darkorange"
}

colors = []
for label, parent in zip(labels, parents):
    if label == root_label:
        colors.append("lightgrey")
    elif label in group_colors:
        colors.append(group_colors[label])   # parent nodes
    elif parent in group_colors:
        colors.append(group_colors[parent])  # child nodes inherit parent color
    else:
        colors.append("steelblue")

# Plot
fig = go.Figure(go.Sunburst(
    labels=labels,
    parents=parents,
    values=values,
    branchvalues="total",
    sort=False,
    insidetextorientation="auto",
    textinfo="text",
    texttemplate=text_template,
    insidetextfont=dict(size=12, color="white"),
    marker=dict(
        colors=colors,
        line=dict(color="white", width=1)
    )
))

fig.update_layout(
    showlegend=False,
    autosize=False,
    width=950,
    height=950,
    template="plotly_white",
    margin=dict(l=10, r=10, t=10, b=10)
)

fig.show()

# fig.write_image(file_name)
```

```python
import plotly.graph_objects as go
import pandas as pd
import textwrap

# === Output file name ===
file_name = "fig_distribution_of_task_classification_by_category.pdf"

# === Data ===
task_classification = [
    "Question Answering",
    "Question Answering; Generation",
    "Model-Efficiency; Inference",
    "Generation",
    "Tracking; Localization",
    "Question Answering; Reasoning",
    "Recognition; Alignment",
    "Recommendation",
    "Retrieval; Classification",
    "Diagnosis; Retrieval",
    "Recognition; Generation",
    "Sentiment Analysis; Emotion-Analysis",
    "Retrieval; Identification",
    "Retrieval; Alignment",
    "Translation",
    "Translation",
    "Representation-Learning",
    "Sentiment Analysis",
    "Forecasting",
    "Classification",
    "Generation; Editing",
    "Representation-Learning",
    "Recognition",
    "Segmentation",
    "Question Answering",
    "Diagnosis",
    "Recognition",
    "Prediction",
    "Matching; Retrieval",
    "Diagnosis; Retrieval",
    "Mapping; Classification",
    "Detection; Classification",
    "Retrieval",
    "Reasoning; Problem-Solving",
    "Detection; Classification",
    "Question Answering",
    "Generation; Recognition",
    "Recognition",
    "Monitoring; Anomaly Detection",
    "Representation-Learning; Generation",
    "Tokenization; Representation-Learning",
    "Understanding; Recognition",
    "Recommendation; Representation-Learning",
    "Understanding; Reasoning",
    "Representation Learning",
    "Grounding; Recognition",
    "Tokenization; Understanding; Generation",
    "Instruction-Tuning; Understanding",
    "Instruction-Following; Understanding",
    "Understanding; Grounding",
    "Conversation; Generation",
    "Multimodal-Modeling; Understanding",
    "Retrieval",
    "Question Answering; Reasoning",
    "Recognition; Reasoning",
    "Recognition",
    "Identification",
    "Sentiment Analysis"
]

task_category = [
    "Single", "Multiple", "Multiple", "Single", "Multiple",
    "Multiple", "Multiple", "Single", "Multiple", "Multiple",
    "Multiple", "Multiple", "Multiple", "Multiple", "Single",
    "Single", "Single", "Single", "Single", "Single",
    "Multiple", "Single", "Single", "Single", "Single",
    "Single", "Single", "Single", "Multiple", "Multiple",
    "Multiple", "Multiple", "Single", "Multiple", "Multiple",
    "Single", "Multiple", "Single", "Multiple", "Multiple",
    "Multiple", "Multiple", "Multiple", "Multiple", "Single",
    "Multiple", "Multiple", "Multiple", "Multiple", "Multiple",
    "Multiple", "Multiple", "Single", "Multiple", "Multiple",
    "Single", "Single", "Single"
]

# === Create DataFrame ===
df = pd.DataFrame({
    "Task Classification": task_classification,
    "Task Category": task_category
})

# Count repeated task classifications within each category
frequency_df = (
    df.groupby(
        ["Task Category", "Task Classification"],
        as_index=False
    )
    .size()
    .rename(columns={"size": "Studies"})
)

# === Helper function for wrapping long labels ===
def wrap_label(text, width=20):
    return "<br>".join(
        textwrap.wrap(
            text,
            width=width,
            break_long_words=False,
            break_on_hyphens=False
        )
    )

# === Build Sunburst hierarchy ===
root_id = "root"
root_label = "Task Classification"
total_studies = len(df)

ids = [root_id]
labels = [root_label]
parents = [""]
values = [total_studies]
display_text = [
    f"<b>{wrap_label(root_label)}</b><br>"
    f"n={total_studies}, 100%"
]
colors = ["#D3D3D3"]

category_colors = {
    "Single": "#4169E1",
    "Multiple": "#2E8B57"
}

child_colors = {
    "Single": "#7895E8",
    "Multiple": "#72B28E"
}

# Task Category is the parent
for category in ["Single", "Multiple"]:
    category_df = frequency_df[
        frequency_df["Task Category"] == category
    ].sort_values(
        by=["Studies", "Task Classification"],
        ascending=[False, True]
    )

    category_total = int(category_df["Studies"].sum())
    category_id = f"category::{category}"

    # Parent category node
    ids.append(category_id)
    labels.append(f"{category} Tasks")
    parents.append(root_id)
    values.append(category_total)
    display_text.append(
        f"<b>{category} Tasks</b><br>"
        f"n={category_total}, "
        f"{category_total / total_studies:.1%}"
    )
    colors.append(category_colors[category])

    # Task Classification children
    for _, row in category_df.iterrows():
        classification = row["Task Classification"]
        study_count = int(row["Studies"])

        child_id = f"{category_id}::{classification}"

        ids.append(child_id)
        labels.append(classification)
        parents.append(category_id)
        values.append(study_count)

        display_text.append(
            f"<b>{wrap_label(classification, width=18)}</b><br>"
            f"n={study_count}, "
            f"{study_count / total_studies:.1%}"
        )
        colors.append(child_colors[category])

# === Create Sunburst chart ===
fig = go.Figure(
    go.Sunburst(
        ids=ids,
        labels=labels,
        parents=parents,
        values=values,
        branchvalues="total",
        sort=False,

        # Wrapped custom labels
        text=display_text,
        textinfo="text",
        insidetextorientation="radial",
        insidetextfont=dict(
            size=11,
            color="white"
        ),

        marker=dict(
            colors=colors,
            line=dict(
                color="white",
                width=1
            )
        ),

        hovertemplate=(
            "<b>%{label}</b><br>"
            "Number of studies: %{value}<br>"
            "Percentage: %{percentRoot:.1%}"
            "<extra></extra>"
        )
    )
)

# === Layout ===
fig.update_layout(
    showlegend=False,
    autosize=False,
    width=1200,
    height=1200,
    template="plotly_white",
    paper_bgcolor="white",
    plot_bgcolor="white",
    margin=dict(l=15, r=15, t=15, b=15),
    uniformtext=dict(
        minsize=8,
        mode="show"
    )
)

fig.show()

```
